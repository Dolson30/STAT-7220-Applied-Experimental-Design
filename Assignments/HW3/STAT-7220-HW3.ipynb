7.  Fit a full factorial ANOVA model to the data. Assess the normality assumption using both a visual method and a testing method. Do we have evidence to support the assumption of normality? Why or why not?

Visual
- Upon inspection, the data points fall tightly along the straight diagonal line. This close alignment suggests that the shape of the residual distribution is approximately Gaussian.

Shapiro-Wilk
- $W = 0.9595$, with a p-value of $0.4291$.Interpretation: Using a standard $\alpha = 0.05$, the p-value ($0.4291$) is substantially greater than the significance level. We therefore fail to reject the null hypothesis.

Because of both of these results, we have strong evidence to support the assumption of normality.



```python
from statsmodels.formula.api import ols
from scipy.stats import shapiro, probplot
import matplotlib.pyplot as plt

# --- 1. Fit the Full Factorial ANOVA Model (Q7 part) ---
model_formula = 'Comfort_Rating ~ A_Sole_Material * B_Insole_Thickness * C_Breathability'
model = ols(model_formula, data=df).fit()

# --- 2. Assess the Normality Assumption (Residuals) ---
residuals = model.resid

print("--- Normality Assumption Check ---")

# A. Visual Method: Q-Q Plot
plt.figure(figsize=(6, 6))
probplot(residuals, dist="norm", plot=plt)
plt.title("Q-Q Plot of Model Residuals")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Standardized Residuals")
plt.show()

# B. Testing Method: Shapiro-Wilk Test
stat, p_value = shapiro(residuals)

print(f"\nShapiro-Wilk Test Statistic (W): {stat:.4f}")
print(f"Shapiro-Wilk p-value: {p_value:.4f}")
```


8.  Assess the constant variance assumption using both a visual method and a testing method. Do we have evidence to support the assumption of constant variance? Why or why not?

Visual:
- The scatter of the points appears to be random and roughly uniform across the range of fitted values (from about 6.5 to 11.0). There is no obvious funnel shape or any clear pattern. Visually, the plot supports the assumption of constant variance.

Levene's
- Using a standard $\alpha = 0.05$, the p-value ($0.4329$) is significantly greater than the significance level. We therefore fail to reject the null hypothesis ($H_0$).

We have strong evidence to support the assumption of constant variance.

```python
from scipy.stats import levene
import matplotlib.pyplot as plt
import statsmodels.api as sm

fitted_values = model.fittedvalues
residuals = model.resid

print("--- Constant Variance Assumption Check ---")

# A. Visual Method: Residuals vs. Fitted Plot
plt.figure(figsize=(8, 5))
plt.scatter(fitted_values, residuals, alpha=0.7)
plt.axhline(0, color='red', linestyle='--', linewidth=1)
plt.title("Residuals vs. Fitted Values Plot (Visual Check)")
plt.xlabel("Fitted Comfort Rating")
plt.ylabel("Residuals")
plt.show()

# --- 2. Testing Method: Levene's Test for Homoscedasticity ---
groups = []
for name, group_data in df.groupby(['A_Sole_Material', 'B_Insole_Thickness', 'C_Breathability']):
    groups.append(group_data['Comfort_Rating'])

stat, p_value = levene(*groups, center='median')

print(f"\nLevene's Test Statistic (W): {stat:.4f}")
print(f"Levene's p-value: {p_value:.4f}")
```


10. Calculate and interpret the partial $\eta^2$ values for each of the main effects and interactions. Which effects and/or interactions are most important? Why?

$\eta^2$ helps us quantify the proportion of the variance in the comfort rating that is uniquely explained by each factor or interaction, controlling for all other effects in the model.

The most important effects are those with the largest partial $\eta^2$ values. According to Cohen's guidelines for effect size, $\eta^2_p$ values around $0.01$ are considered small, $0.06$ are medium, and $0.14$ are large.

- A_Sole_Material - $\mathbf{0.2625}$ - Large
- B_Insole_Thickness:C_Breathability - $0.0609$ - Medium
- B_Insole_Thickness - $0.0406$ - Small
- A_Sole_Material:B_Insole_Thickness:C_Breathability- $0.0394$ - Small
- A_Sole_Material:B_Insole_Thickness- $0.0190$ - Small
- C_Breathability - $0.0040$ - Small
- A_Sole_Material:C_Breathability - $0.0000$ - Small/Negligible Effect


```python
from statsmodels.stats.anova import anova_lm
import pandas as pd

# --- 1. Get the ANOVA Table Data ---
anova_table = anova_lm(model, typ=3)
ss_error = anova_table.loc['Residual', 'sum_sq']

# --- 2. Calculate Partial Eta Squared (η²p) ---
eta_squared_p = {}

for effect in anova_table.index:
    if effect not in ['Intercept', 'Residual']:
        ss_effect = anova_table.loc[effect, 'sum_sq']

        partial_eta2 = ss_effect / (ss_effect + ss_error)
        eta_squared_p[effect] = partial_eta2

eta_df = pd.DataFrame(list(eta_squared_p.items()), columns=['Effect', 'Partial_Eta2'])
eta_df.set_index('Effect', inplace=True)
eta_df.sort_values(by='Partial_Eta2', ascending=False, inplace=True)

print("\n--- Partial Eta Squared (η²p) Values ---")
print(eta_df.to_string(float_format='%.4f'))
```

4.  Explain the difference between a full $2^k$ factorial design and a $2^{3-1}$ fractional factorial design. Briefly comment on why the latter may be preferred in this case.
- A full $2^k$ factorial design tests every possible combination of factor levels, requiring $2^3 = 8$ unique runs to study the three factors (A, B, C). This design allows for the clear and independent estimation of all main effects and all interaction effects. In contrast, a $2^{3-1}$ fractional factorial design uses only half the runs, resulting in $2^{3-1} = 4$ total runs for this experiment.  This reduced number of runs means that effects become aliased, or confounded; for example, the estimated main effect of Factor A is inseparable from the two-way interaction of B and C ($A \equiv BC$).
- The fractional design is preferred here because the company faces time and resource limitations. By accepting the risk of aliasing—based on the common assumption that high-order interactions are negligible, the manufacturer can use half the resources to efficiently screen the factors.

6.  Assuming a defining relation of $I=ABC$, specify the aliasing structure of the design. What does this mean for the interpretation of the main effects and interactions?

Main Effect Aliases:
- $A \cdot (I=ABC) \implies A = A^2BC = BC$
- $B \cdot (I=ABC) \implies B = AB^2C = AC$
- $C \cdot (I=ABC) \implies C = ABC^2 = AB$

Resulting Aliases:
- $A \equiv BC$
- $B \equiv AC$
- $C \equiv AB$

This aliasing means that the value calculated for a main effect, such as $A$ (GPS Accuracy), is actually the sum of the true effect of $A$ and the true effect of the $BC$ (Battery Life $\times$ Display) interaction. Because the experimental design sacrifices independent estimation, we must rely on the effect sparsity principle, which assumes that the higher-order interactions are negligible

12. Calculate and interpret the partial $\eta^2$ values for each of the main effects. Which effects are most important? Why?

As an example: The Partial $\eta^2$ value of $0.9545$ indicates that Display Size accounts for approximately $95.45\%$ of the variance in satisfaction, after accounting for all other effects in the model.

C (Display) ≡ AB, 0.9545, 1st

A (GPS) ≡ BC, 0.9312, 2nd

B (Battery) ≡ AC, 0.7087, 3rd

Display is the most important due to the largest $\eta^2$ value, with GPS in second followed by Battery. In practical terms, making changes to the Display Size or the GPS Accuracy level has the most significant, singular impact on a user's satisfaction rating, far exceeding the contribution of Battery Life.

```python
mse = model.mse_resid
df_residual = model.df_resid
num_observations = model.nobs

coefficients = model.params.drop('const')
t_stats = model.tvalues.drop('const')

ss_effect = (t_stats ** 2) * mse * 1
ss_error = model.ssr

partial_eta_squared = ss_effect / (ss_effect + ss_error)

eta_df = partial_eta_squared.reset_index()
eta_df.columns = ['Effect', 'Partial_Eta2']

aliases = [
    'A (GPS) ≡ BC',
    'B (Battery) ≡ AC',
    'C (Display) ≡ AB'
]
eta_df['Alias'] = aliases
eta_df['Importance Rank'] = eta_df['Partial_Eta2'].abs().rank(ascending=False).astype(int)

eta_df = eta_df[['Alias', 'Partial_Eta2', 'Importance Rank']]

print("\n--- Partial Eta Squared (η²p) Values (Q12) ---")
print(eta_df.to_markdown(index=False, floatfmt=".4f"))
print("-" * 50)
```
